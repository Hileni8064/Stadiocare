```
import pandas as pd

df = pd.read_csv(
    "data/processed/cleaned_data.csv"
)

# Date conversion
df["admittime"] = pd.to_datetime(
    df["admittime"]
)

df["dischtime"] = pd.to_datetime(
    df["dischtime"]
)

# Target Variable
df["length_of_stay"] = (
    df["dischtime"] -
    df["admittime"]
).dt.days

# Time Features
df["admission_month"] = (
    df["admittime"].dt.month
)

df["day_of_week"] = (
    df["admittime"].dt.dayofweek
)

df["weekend_admission"] = (
    df["day_of_week"] >= 5
).astype(int)

# One-Hot Encoding
categorical_features = []

for col in [
    "admission_type",
    "insurance",
    "race"
]:
    if col in df.columns:
        categorical_features.append(col)

df = pd.get_dummies(
    df,
    columns=categorical_features,
    drop_first=True
)

df.to_csv(
    "data/processed/engineered_data.csv",
    index=False
)

print("Feature engineering complete")
print(
    "Saved to: data/processed/engineered_data.csv"
)
```
