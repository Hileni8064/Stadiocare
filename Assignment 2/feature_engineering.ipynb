```
# Feature Engineering Notebook

import pandas as pd

# Load preprocessed dataset
df = pd.read_csv("data/FreeInsuranceClaims100M_sample.csv")

# Example engineered features
df["claim_tenure_ratio"] = df["claim_amount"] / (df["tenure"] + 1)
df["previous_claims_bucket"] = pd.cut(df["previous_claims_count"], bins=[0,1,3,5,10], labels=[0,1,2,3])
df["incident_month"] = pd.to_datetime(df["incident_date"]).dt.month
df["incident_year"] = pd.to_datetime(df["incident_date"]).dt.year
df["deductible_claim_interaction"] = df["deductible"] * df["claim_amount"]

print("Feature engineering complete. New columns added:", 
      ["claim_tenure_ratio", "previous_claims_bucket", "incident_month", "incident_year", "deductible_claim_interaction"])
```
