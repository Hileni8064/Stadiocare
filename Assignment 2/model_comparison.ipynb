```
# ==========================================
# MODEL COMPARISON
# Random Forest vs XGBoost
# ==========================================

import pandas as pd

print("=" * 50)
print("MODEL COMPARISON")
print("=" * 50)

rf_mae = float(
    input("Random Forest MAE: ")
)

rf_rmse = float(
    input("Random Forest RMSE: ")
)

rf_r2 = float(
    input("Random Forest R²: ")
)

xgb_mae = float(
    input("XGBoost MAE: ")
)

xgb_rmse = float(
    input("XGBoost RMSE: ")
)

xgb_r2 = float(
    input("XGBoost R²: ")
)

results = pd.DataFrame({
    "Metric": [
        "MAE",
        "RMSE",
        "R²"
    ],
    "Random Forest": [
        rf_mae,
        rf_rmse,
        rf_r2
    ],
    "XGBoost": [
        xgb_mae,
        xgb_rmse,
        xgb_r2
    ]
})

print("\nResults Summary")
print(results)

print("\nBest Performing Model")

if xgb_r2 > rf_r2:
    print("XGBoost Regression")
else:
    print("Random Forest Regression")
```
